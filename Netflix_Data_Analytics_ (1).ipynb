```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
df = pd.read_csv("netflix_titles.csv")
df.head()
```

```python
df.shape
df.columns
df.info()
df.describe()
```

#Data Cleaning

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df['director'].fillna('Unknown', inplace=True)
df['cast'].fillna('Not Available', inplace=True)
df['country'].fillna('Unknown', inplace=True)
df['rating'].fillna(df['rating'].mode()[0], inplace=True)
df['duration'].fillna(df['duration'].mode()[0], inplace=True)
```

#Content Type (Movies vs TV Shows)

```python
sns.countplot(x='type', data=df)
plt.title("Movies vs TV Shows")
plt.show()
```

#Content by Year

```python
df['release_year'].value_counts().sort_index().plot()
plt.title("Content Release Trend Over Years")
plt.show()
```

#Top Countries

```python
df['country'].value_counts().head(10).plot(kind='bar')
plt.title("Top Countries Producing Content")
plt.show()
```

#Genre Analysis


```python
df['listed_in'].value_counts().head(10).plot(kind='barh')
plt.title("Top Genres")
plt.show()
```

#Ratings Distribution


```python
sns.countplot(y='rating', data=df, order=df['rating'].value_counts().index)
plt.title("Content Ratings")
plt.show()
```

#Duration Analysis

```python
df['duration'].value_counts().head(10).plot(kind='bar')
plt.title("Duration Distribution")
plt.show()
```

#Correlation Heatmap

```python
plt.figure(figsize=(8,5))
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

#Year vs Type

```python
sns.countplot(x='release_year', hue='type', data=df)
plt.xticks(rotation=90)
plt.title("Movies vs TV Shows Over Years")
plt.show()
```

#Top Directors


```python
df['director'].value_counts().head(10).plot(kind='bar')
plt.title("Top Directors")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```
